# Creating a manual model using the pairplots
Tring to create a visual manual analysis of the Iris flower data using the pairplots and histograms. 
The baseline of this model is 0.33 or 33% as it's a balanced dataset with 50 flowers in each species of three total species summed to get a total of 150 flowers. 

```python
def manual_predictor(petel_length):
    if petel_length < 2.5:
        return 0
    elif petel_length < 4.9:
        return 1
    else:
        return 2
```

```python
df_train.columns
```

```python
x_train[:, 2]
```

```python
manual_prediction_result = [manual_predictor(val) for val in x_train[:, 2]]
```

```python
y_train
```

```python
accuracy = np.mean(manual_prediction_result == y_train)
```

```python
print(f'Accuracy of a manual model: {accuracy * 100:.2f}')
```

### Cross Validating the evaluated model

```python
from sklearn.model_selection import cross_val_predict, cross_val_score
```

```python
model = LogisticRegression(max_iter = 200)
```

```python
collective_accuracies = cross_val_score(model, x_train, y_train, cv = 5, scoring = "accuracy")
```

```python
collective_accuracies
```

```python
accuracy = np.mean(collective_accuracies)
print(f"Accuracy: {accuracy * 100:.2f}%")
```

```python
cross_y_prediction = cross_val_predict(model, x_train, y_train, cv = 5)
```

```python
correct_prediction = cross_y_prediction == y_train
```

```python
new_df_train = df_train.copy()
```

```python
new_df_train["cross_y_pred"] = cross_y_prediction
```

```python
new_df_train["right_pred"] = correct_prediction
```

```python
new_df_train["right_pred_label"] = new_df_train["cross_y_pred"].map({0: "setosa", 1: "versicolor", 2: "virginica"})
```

```python
sns.scatterplot(x  = "petal length (cm)", y= "petal width (cm)", hue = "right_pred_label", data= new_df_train)
```

```python
sns.scatterplot(x  = "petal length (cm)", y= "petal width (cm)", hue = "names", data= new_df_train)
```

```python
def incorrect_predictions(df, x_axis, y_axis):
    fig, axs = plt.subplots(2, 2, figsize = (8, 8))
    axs = axs.flatten()
    sns.scatterplot(x = x_axis, y = y_axis, hue = "right_pred_label", data = new_df_train, ax = axs[0])
    sns.scatterplot(x = x_axis, y = y_axis, hue = "names", data = new_df_train, ax = axs[1])
    sns.scatterplot(x = x_axis, y = y_axis, hue = "right_pred", data = new_df_train, ax = axs[2])
    axs[3].set_visible(False)
```

```python
incorrect_predictions(new_df_train, "petal length (cm)", "petal width (cm)")
```

## Model Tuning

```python
for param in (0.5, 1, 2, 7, 8, 9):
    print(param)
    model = LogisticRegression(max_iter = 250, C = param)
    model.fit(x_train, y_train)
    tuned_accuracy = np.mean(cross_val_score(model, x_train, y_train, cv = 5, scoring = "accuracy"))
    print(f"Accurcy for param: {tuned_accuracy * 100:.2f}%")
```

```python
x_test = df_test.drop(columns = ["target", "names"]).values
y_test = df_test["target"]
```

```python
model = LogisticRegression(max_iter = 300, C = 2)
model.fit(x_train, y_train)
tuned_accuracy = np.mean(cross_val_score(model, x_test, y_test, cv = 5, scoring = "accuracy"))
print(f"Tuned accuracy: {tuned_accuracy * 100:.2f}%")
```

```python
new_test_df = df_test.copy()
new_test_df["y_test"] = y_test
new_test_df["y_pred"] = y_pred
new_test_df["result"] = res
```

```python
new_test_df
```

```python
def plotting_incorrect_data(df, x_axis, y_axis):
    fig, axs = plt.subplots(2, 2, figsize = (8, 8))
    axs = axs.flatten()
    sns.scatterplot(x = x_axis, y = y_axis, hue = "y_test", data = new_test_df, ax = axs[0])
    sns.scatterplot(x = x_axis, y = y_axis, hue = "y_pred", data = new_test_df, ax = axs[1])
    sns.scatterplot(x = x_axis, y = y_axis, hue = "result", data = new_test_df, ax = axs[2])
    axs[3].set_visible(False)
```

```python
plotting_incorrect_data(new_test_df, "petal length (cm)", "petal width (cm)")
```
